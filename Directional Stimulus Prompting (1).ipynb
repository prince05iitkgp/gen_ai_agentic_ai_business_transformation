```python
example_prompt=PromptTemplate(
    input_variables=["Article", "Keywords", "Summary"],
    template="""
    Article: {Article}
    Q: Given an article and a list of keywords, write a short summary that accurately incorporates the provided keywords into 2-4 sentences.
    Keywords: {Keywords}
    A: {Summary}
    """
)
```

```python
one_shot_prompt = FewShotPromptTemplate(
    examples=example,
    example_prompt=example_prompt,
    suffix="""
    Article: {input}
    Q: Given an article and a list of keywords, write a short summary that accurately incorporates the provided keywords into 2-4 sentences.
    Keywords: {keywords}
    A:
    """,
    input_variables=["input", "keywords"],
)
```
