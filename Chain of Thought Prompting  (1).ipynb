```python
prompt_no_cot = """
A company invests ₹2 crore in AI automation and saves ₹80 lakh per year.
Is this a good decision?
"""
```

```python
prompt_with_cot = """
A company invests ₹2 crore in AI automation and saves ₹80 lakh per year.

Think step by step:
1. Calculate payback period
2. Analyze financial feasibility
3. Conclude whether the investment is justified

Provide reasoning before the final answer.
"""
```

```python
from openai import OpenAI
import os

client = OpenAI()

def run_prompt(prompt):
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": prompt}],
        temperature=0
    )
    return response.choices[0].message.content


print("WITHOUT CHAIN OF THOUGHT:\n")
print(run_prompt(prompt_no_cot))

print("\n" + "="*80 + "\n")

print("WITH CHAIN OF THOUGHT:\n")
print(run_prompt(prompt_with_cot))
```
